```python
# ==============================================================================
# PROGRAM: Student Profile
# TOPIC: Data Types - Student Profile
# AUTHOR: Sarosh Khan
# DATE: August 2026
# ==============================================================================

# ==============================================================================
# STEP 1: Variables Declaration
# ==============================================================================

name = "Sarosh Khan"          # str - Student's name
age = 25                      # int - Student's age
gpa = 3.8                     # float - Student's GPA
is_active = True              # bool - Student's active status

# ==============================================================================
# STEP 2: Display Header
# ==============================================================================

print("\n" + "=" * 50)
print("          STUDENT PROFILE")
print("=" * 50 + "\n")

# ==============================================================================
# STEP 3: Display Student Information with Data Types
# ==============================================================================

print("STUDENT INFORMATION")
print("-" * 40)
print(f"Name    : {name}    -> {type(name)}")
print(f"Age     : {age}     -> {type(age)}")
print(f"GPA     : {gpa}     -> {type(gpa)}")
print(f"Active  : {is_active}    -> {type(is_active)}")

# ==============================================================================
# STEP 4: Type Conversion Section
# ==============================================================================

print("\n" + "=" * 50)
print("       TYPE CONVERSION")
print("=" * 50 + "\n")

# Perform type conversion
age_str = str(age)           # int → str
gpa_int = int(gpa)           # float → int
active_str = str(is_active)  # bool → str

# Display converted values
print("CONVERTED VALUES")
print("-" * 40)
print(f"Age as String  : {age_str} -> {type(age_str)}")
print(f"GPA as Integer : {gpa_int} -> {type(gpa_int)}")
print(f"Active as String: {active_str} -> {type(active_str)}")

# ==============================================================================
# STEP 5: Program End
# ==============================================================================

print("\n" + "=" * 50)
print("          PROGRAM END")
print("=" * 50 + "\n")
```
