Problem Statement:

Write a python Program that calculate simple interest.

Formula: Simple Interest = (principle x Rate x Time) / 100

Input:
------
1.Principle Amount (float) initial amount of money
2.Rate of Interest (float) The interest rate per year
3.Time Period (float) The times in years

Output:
-------
1. Simple Interest calculated.
2. Total  Amount = Principe + Simple Interest

STEP-BY-STEP LOGICAL FLOW:
--------------------------
1. Start the program

2. Create 3 variables:
   - principal_amount = 0.0 (float)
   - rate_of_interest = 0.0 (float)
   - time_period = 0.0 (float)

3. Display Header:
   print("=" * 50)
   print("       SIMPLE INTEREST CALCULATOR")
   print("=" * 50)

4. Get user input:
   - principal_amount = float(input("Enter Principal Amount: "))
   - rate_of_interest = float(input("Enter Rate of Interest: "))
   - time_period = float(input("Enter Time Period (years): "))

5. Calculate Simple Interest:
   simple_interest = (principal_amount * rate_of_interest * time_period) / 100

6. Calculate Total Amount:
   total_amount = principal_amount + simple_interest

7. Display Results Header:
   print("\n" + "=" * 50)
   print("              RESULTS")
   print("=" * 50)

8. Display Results with Data Types:
   - print(f"Principal    : {principal_amount} -> {type(principal_amount)}")
   - print(f"Rate         : {rate_of_interest} -> {type(rate_of_interest)}")
   - print(f"Time         : {time_period} -> {type(time_period)}")
   - print(f"Simple Interest : {simple_interest} -> {type(simple_interest)}")
   - print(f"Total Amount : {total_amount} -> {type(total_amount)}")

9. Display Footer:
   print("=" * 50 + "\n")

10. End the program


```python
#step 1: Create three variables and initialize 0
principle_amount = 0.0
rate_of_interest = 0.0
time_period = 0.0

#step 2: Take input from user for principle amount, rate of interest and time period
print("\n" + "=" * 50)
print("\n"         "SIMPLE INTEREST CALCULATOR" + "\n")
print("=" * 50)

#step 3: Take input from user for principle amount, rate of interest and time period
principle_amount = float(input("Enter principle amount: "))
rate_of_interest = float(input("Enter rate of interest (in percentage): "))
time_period = float(input("Enter time period (in years): "))

#step 4: Calculate simple interest using the formula: SI = (P * R * T) / 100

simple_interest = (principle_amount * rate_of_interest * time_period) / 100
total_amount = principle_amount + simple_interest
#step 5: Display the calculated simple interest
print("\n" + "=" * 50)
print(f"Simple Interest: {simple_interest} -> type: {type(simple_interest)}")
print(f"Total Amount: {total_amount} -> type: {type(total_amount)}")

#step 6: Display the footer
print("=" * 50 + "\n")
print("Thank you for using the Simple Interest Calculator!")
print("=" * 50 + "\n")
```
