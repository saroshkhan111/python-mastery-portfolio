```python
#step 1: Create and initialize variables

person_name =  "Sarosh Khan"   # str
person_age  =     25   # int
person_height =  5.9   # float
is_student   =  True   # boolean
person_hobbies = ['Reading',  'Coding',  'Gaming']  # list-type
person_address = {
    "Country": "Pakistan",
    "City": "Karachi",
    "Street": "MainStreet",
    "PostalCode": 75000
} # dict type

unique_ids = {101, 102,  103} # step

#step 2: Display their values and data-types

print("\n"  + "=" * 50)
print("\n"         "PERSON'S   INFORMATION")
print("\n"  + "=" * 50)


print(f"Name    :   {person_name}-> {type(person_name)}\n")

print(f"Age     :   {person_age}-> {type(person_age)}\n")

print(f"Height  :   {person_height}->   {type(person_height)}\n")

print(f"Student :   {is_student}->   {type(is_student)}\n")

print(f"Hobbies :   {person_hobbies}->   {type(person_hobbies)}\n")

print(f"Address :   {person_address}->   {type(person_address)}\n")

print(f"Unique_ids :   {unique_ids}->     {type(unique_ids)}\n")


#step 3: Convert the data-types in the different formats;

print("\n"  + "=" * 50)
print("\n"              "TYPE CONVERSION")
print("\n"  + "=" * 50)


age_convert = str(person_age) # int to str
height_convert = int(person_height) # float to int
student_convert = str(is_student) # boolean to str
hobbies_convert = tuple(person_hobbies) # list to tuple


print(f"Age  as string :  {age_convert}-> {type(age_convert)}\n")
print(f"Height as integer : {height_convert}-> {type(height_convert)}\n")
print(f"Student as string : {student_convert}-> {type(student_convert)}\n")
print(f"Hobbies as tuple  : {hobbies_convert}-> {type(hobbies_convert)}\n")

print("\n"  + "=" * 50)
print("\n"              "PROGRAM END")
print("\n"  + "=" * 50)
```
