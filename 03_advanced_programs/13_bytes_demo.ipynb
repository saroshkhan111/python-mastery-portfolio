Definition:
-----------

Bytes are the raw digital data that computers understand. They are a sequence of numbers (0 - 255)
that represent information like text, images, or files

Simple Explanation:
-------------------

Think of bytes as the computer's native language. While humans read text like "Hello", 
computers read it as bytes: b'Hello' or [72, 101, 108, 108, 111].

Key Points:
-----------
1. Immutable: Cannot be changed after creation.

2. Sequence: Ordered collection of integers (0-255).

3. Syntax: Written with a b prefix, e.g., b"Hello".

4. Used for: Files, network communication, images, cryptography.

5. Memory Efficient: Takes less space than strings

6. Convertible: Can be changed to/from strings using .encode() and .decode().

Use Cases in AI/Programming:
----------------------------

*Reading image files (open("image.jpg", "rb"))

*Sending data over network (socket.send(data))

*Cryptography/hashing (hashlib.md5(b"password"))

*Working with binary files (PDF, ZIP, etc.)
