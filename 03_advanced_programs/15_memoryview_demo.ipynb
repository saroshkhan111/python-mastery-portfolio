Concept of Memory View data-type in python:
-------------------------------------------

Binary Object Concept:
----------------------

jise aam tor par BLOB ya Binary Large Object bhi kaha jata hai 
raw data ka aik aisa majmua hota hai jo computer mein 0 aur 1 (binary form) mein store hota hai.
Yeh aam tor par aisi files hoti hain jinhein normal text ki tarah parha nahi ja sakta
balkay inhein chalane ya dekhne ke liye kisi khas software ki zaroorat hoti hai

🔎 Binary Objects Ki Misaalein (Examples)
------------------------------------------
*Multimedia Files: Photos (JPEG, PNG), Audio (MP3, WAV), aur Videos (MP4, MKV).
*Documents: PDFs aur Microsoft Word files
*Software: Executable files (.exe ya .app) aur compiled code.
*Compressed Files: ZIP ya RAR archives

Memory View Kya Hai?:
---------------------

*Memory view Python ka ek built-in data type hai
*Ye aapko binary objects ki memory ko directly access karne deta hai
*Ye original data ki koi COPY nahi banata
*Isko ek "khidki" ki tarah samjho jo same data ko dekhti hai

Hume Iska Zaroorat Kyun Hai?:
-----------------------------

*Normal slicing memory mein data ki NAYI copy banati hai
*Memory view koi copy nahi banata — ye sirf original data ko point karta hai
*Ye bade datasets ke saath kaam karte waqt bahut saari RAM bachata hai
*Ye aapke program ko TEZI se chalata hai

Kaunse Objects Memory View Support Karte Hain?:
-----------------------------------------------

*bytes (immutable — tabdeel nahi ho sakta)
*bytearray (mutable — tabdeel ho sakta hai)
*array.array
*numpy arrays

Khas Khusoosiyaat:
-----------------

*Indexing aur slicing support karta hai
*.tobytes() se wapas bytes mein convert kar sakte ho
*.tolist() se wapas list mein convert kar sakte ho
*.release() se kaam khatam hone par memory free kar sakte ho
*.cast() se data ko alag formats mein reshape kar sakte ho

Zaroori Qanoon:
---------------

*Agar original object mutable hai (jaise bytearray)
*Memory view ke zariye kiye gaye changes original object ko BHI affect karenge
*Dono ek hi SAME memory location ko point karte hain

```python
#step 1: Print Header and line with MEMORYVIEW OPERATIONS DEMONSTRATION

print("\n" + "=" * 50)
print("\n...............MEMORYVIEW OPERATIONS DEMONSTRATION")
print("\n" + "=" * 50)


#step 2: Create a byte  and bytearray 
mv_data = b'Hello AI World'
mv_ba   = bytearray(b'Python')

#step 3: Print the original byte and bytearray with their types
print(f"\nBYTES: {mv_data} TYPE: {type(mv_data)}\n")
print(f"\nBYTEARRAY: {mv_ba} TYPE: {type(mv_ba)}\n")

#step 4: Convert the byte and bytearray to memoryview and print them with their types
mv_data_view = memoryview(mv_data)
mv_ba_view  =  memoryview(mv_ba)

print(f"\nMEMORYVIEW of BYTES: {mv_data_view} TYPE: {type(mv_data_view)}\n")
print(f"\nMEMORYVIEW of BYTEARRAY: {mv_ba_view} TYPE: {type(mv_ba_view)}\n")

#step 5: Access the elements of the memoryview and print them
print(f"\nFIRST  ELEMENT : {mv_data_view[0]} -> {type((mv_data_view[0]))}\n")
print(f"\nSECOND ELEMENT : {mv_data_view[1]} -> {type(mv_data_view[1])}\n")
print(f"\nLAST   ELEMENT : {mv_data_view[-1]}-> {type(mv_data_view[-1])}\n")

#step 6: Access elements using slicing:
print(f"\nACCESSING 1st 5 ELEMENT using SLICING: {mv_data_view[0:5]}\n")

#step 7: print the length of mv_data and mv_ba using len() function
print(f"\nLENGHT OF BYTES: -> {len(mv_data)}\n")
print(f"\nLENGTH OF BYTEARRAY: -> {len(mv_ba)}\n")

#step 8: COnvert mv_data in bytes
mv_bytes = mv_data_view.tobytes()
print(f"\nAfter converting MEMORYVIEW in BYTES {mv_bytes} -> {type(mv_bytes)}\n")

#step 9: Cpnvert mv_data in list
mv_list = mv_data_view.tolist()
print(f"\nAfter Conversion of MEMORYVIEW in LIST {list(mv_list)} -> {type(mv_list)}\n")

#step 10: Convert mv_data in tuple
mv_tuple = tuple(mv_data_view)


print(f"\nmv_data.readonly: {mv_data_view.readonly}")   # True (bytes)
print(f"mv_ba_view.readonly: {mv_ba_view.readonly}")   # False (bytearray)

print(f"\nTuple Conversion of MEMORYVIEW: {tuple(mv_tuple)} -> {type(mv_tuple)}\n")

#step 11: Print Each Element Using For loop
print("\nPrinting Each Element of MEMORYVIEW using FOR LOOP\n")
for i , j in enumerate(mv_data_view):
    print(f"Element {i} : {j}")

#step 12: Release the memoryview objects
mv_data_view.release()
mv_ba_view.release()

print(f"\nAfter Release mv_data_view and mv_ba_view -> {mv_data_view} and {mv_ba_view}\n")

#step 13: Print Footer
print("\n" + "=" * 70)
print("\n...............MEMORYVIEW OPERATIONS DEMONSTRATION COMPLETED")
print("\n" + "=" * 70)
```
