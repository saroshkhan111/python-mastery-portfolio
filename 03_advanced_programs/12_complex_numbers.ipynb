1. Real part (normal number)

Syntax: real + imaginary j:

   c1.imag  -> 4.0

```python
#st 1: create complex number directly

c1 = 3 + 4j

#st 2: create complex number with complex function()

c2 = complex(5, -2)

#Accessing and print the real part of the complex number
print("\n" + "=" * 50)
print("\n................PRINT REAL AND IMAG PART OF COMPLEX NUMBER:")
print("\n" + "=" * 50)

print(f"\n REAL PART : -> {c1.real}\n")

print(f"\n IMAG PART : -> {c1.imag}\n")


print(f"\n REAL PART : -> {c2.real}\n")

print(f"\n IMAG PART : -> {c2.imag}\n")

#st 3: Perform Arithemtic Operations on Complex number:

print("\n" + "=" * 50)
print("\n................ARITHEMETIC OPERATIONS ON COMPLEX NUMBER:")
print("\n" + "=" * 50)

print(f"\n Add : -> {c1 + c2}\n")

print(f"\n Sub : -> {c1 - c2}\n")


print(f"\n MULT : -> {c1 * c2}\n")

print(f"\n DIV : ->  {c1 / c2}\n")

print("\n" + "=" * 50)
print("\n................USE  CONJUGATE() ON COMPLEX NUMBER:")
print("\n" + "=" * 50)


complex_conjugate = c1.conjugate()

print(f"\nConjugate c1 : -> {complex_conjugate}\n")

complex_magintude = abs(c2)


print("\n" + "=" * 50)
print("\n................USE  ABS() ON COMPLEX NUMBER:")
print("\n" + "=" * 50)

print(f"\nMagintude c2 : -> {complex_magintude}\n")

#st 4: Display all Result with Data-types

print("\n" + "=" * 50)
print("\n................USE  ABS() ON COMPLEX NUMBER:")
print("\n" + "=" * 50)

print(f"\nType of  c1   ->   {type(c1)}\n")

print(f"\nType of  c2   ->   {type(c2)}\n")

print(f"\ntype of  c1 + c2  -> {type(c1 + c2)}\n")

print(f"\ntype of  c1 - c2  -> {type(c1 - c2)}\n")

print(f"\ntype of  c1 * c2  -> {type(c1 * c2)}\n")

print(f"\ntype of  c1 / c2  -> {type(c1 / c2)}\n")


print(f"\ntype of  Conjugate  -> {type(c1.conjugate())}\n")
print(f"\ntype of  abs  -> {type(abs(c1))}\n")
```
