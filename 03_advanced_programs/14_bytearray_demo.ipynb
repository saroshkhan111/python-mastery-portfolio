Concept of byte array:
----------------------

Bytearray = Bytes ki ek changeable (mutable) list. Jaise bytes, lekin isko change kar sakte hain!

Bytes = Read-only (sirf dekh sakte hain)
Bytearray = Read-Write (dekh bhi sakte hain, change bhi kar sakte hain)

Syntax:
# Bytes (Immutable - Change nahi kar sakte)

# b [0]  -> 72 Error

# Bytearray (Mutable - Change kar sakte hain)

ba = bytearray(b'Hello')

ba[0] = 72  changed allowed

```python
# Step 1: Print the Header
print("\n" + "=" * 50)
print("\n.........BYTEARRAY OPERATIONS DEMONSTRATION")
print("\n" + "=" * 50)

# Step 2: Create bytearray from string
ba1 = bytearray('Hello', 'utf-8')   # ✅ Fixed: No space
print(f'\nOriginal BYTE ARRAY = {ba1}')

# Step 3: Create bytearray from bytes
ba2 = bytearray(b'World')

# Step 4: Access byte through index
print(f'\nba1[0] = {ba1[0]} value -> {ba1} Type -> {type(ba1)}\n')

# Step 5: Change byte at index 0 (mutability)
ba1[0] = 63
print(f'\nAFTER CHANGE index[0] = {ba1}')

# Step 6: Append byte
ba1.append(33)
print(f'\nAFTER Append ba1 = {ba1}')

# Step 7: Extend with bytes
ba1.extend(b'!')
print(f'\nAFTER Extend ba1 = {ba1}')

# Step 8: Insert byte at index 0
ba1.insert(0, 72)
print(f"\nAfter insert -> {ba1}")

# Step 9: Pop last item
print(f"\nPop Last Item -> {ba1.pop()}")
print(f"After pop -> {ba1}")

# Step 10: Remove specific byte
ba1.remove(33)   # 
print(f"\nAfter remove -> {ba1}")

# Step 11: Length of bytearrays
print(f"\nLength of bytearray 1: -> {len(ba1)}")   
print(f"\nLength of bytearray 2: -> {len(ba2)}")   

# Step 12: Decode to string
decode_string = ba1.decode()
print(f"\nDecoded string: -> {decode_string} -> {type(decode_string)}")   

# Step 13: Convert to bytes
convert_byte = bytes(ba1)
print(f"\nAfter Converting in Bytes: -> {convert_byte} -> {type(convert_byte)}")


# step 14: Loop Through ba1
print("\nLoop Through ba1\n")
for item in ba1:
    print(item)

# step 15: Slice Bytearray

print(f"\nSlice of Byte Array [0:5] -> {ba1[0:5]}")

# step 16: Clearn bytes array ba2

ba2.clear()
print(f"\nAfter clear ba2 -> {ba2}\n")


print("\n" + "=" * 50)
print("PROGRAM 14 COMPLETED SUCCESSFULLY!")
print("=" * 50)
```
