displays it with a message.
Input:
• Birth year (integer)
Output:

Expected Output:
Enter your birth year: 2000
You are 24 years old.

## Problem Solving Steps:

## Step 1: Problem Statement:

## Problem: Age Calculator
Ask the user about its birth year, and then calculate their age. After that, display it on the screen.

## What is the input?

- Birth year
- Example: 2000, 1995, 1980 etc.
- Take input from the user as a string and convert it into integer format.

## What is the Output?

- Calculate the user's age
- Format: "You are 24 years old."

## What are the conditions?

- Birth year must be less than current year.
- Birth year must be greater than 0.
- Birth year must be an integer number (no text).
- User must be born already (no future date).



## Step 2: Requirements Analysis:

## 1. Tools We Need (Kya Tools Chahiye?)

## 2. Data We Need (Kya Data Chahiye?)

### Output Data:
- "You are X years old."

## 3. Skills We Need:

3. Type Conversion (string to integer)
4. Import library: `import datetime`

6. f-string for formatting: `f"You are {age} years old"`
7. try/except block for error handling (so program does not crash)


### Step 3: Algorithm Design:

START

1. Import datetime library

2. Get the current year from the system
   (current_year = datetime.datetime.now().year)

3. Display message to user
   ("Enter your birth year: ")

4. Get input from the user
   (birth_year = user_input)

5. Validate the user's input:
   - If birth_year is text → "Cannot be text! Enter number."
   - If birth_year is negative → "Only positive numbers allowed!"
   - If birth_year is decimal (float) → "No decimals allowed! Enter integer."
   - If birth_year is in future → "Future year is not allowed!"

6. If all validations pass:
   - Calculate the age
   - Formula: age = current_year - birth_year

7. Display the final result
   - Print: "You are X years old"

END

### Pseudocode:

BEGIN CalculateAge

    IMPORT datetime

    SET current_year = GET_CURRENT_YEAR()
    SET valid_input = FALSE

    WHILE valid_input IS FALSE DO
        DISPLAY "Enter your birth year: "
        READ user_input

        IF IS_INTEGER(user_input) IS FALSE THEN
            IF IS_FLOAT(user_input) IS TRUE THEN
                DISPLAY "No decimals allowed! Enter integer."
            ELSE
                DISPLAY "Cannot be text! Enter number."
            ENDIF
        ELSE
            SET birth_year = CONVERT_TO_INTEGER(user_input)

            IF birth_year < 0 THEN
                DISPLAY "Only positive numbers allowed!"
            ELSE IF birth_year > current_year THEN
                DISPLAY "Future year is not allowed!"
            ELSE
                SET valid_input = TRUE
            ENDIF
        ENDIF
    ENDWHILE

    COMPUTE age = current_year - birth_year
    DISPLAY "You are " + age + " years old"

END

```python
import datetime

# Step 2: Get current year
current_year = datetime.datetime.now().year

# Step 3: Get user input
try:
    birth_year = int(input("Enter your birth year: "))
    
    # Step 4: Validations
    if birth_year <= 0:
        print("Error: Birth year cannot be negative!")
    elif birth_year > current_year:
        print("Error: Future year not allowed!")
    else:
        # Step 5: Calculate
        age = current_year - birth_year
        
        # Step 6: Display
        print(f"You are {age} years old")

except:
    print("Error: Please enter a valid number!")
```
