### Program 3: Simple Interest Calculator (Basic)

##  Problem Statement: Write a program that calculates simple interest.
   - Formula: Simple Interest = (Principal × Rate × Time) / 100


Expected Output:
Enter Principal: 10000
Enter Rate: 5
Enter Time: 2
Simple Interest is: 1000.0   

### What is the Problem:
I need to write a program which will calculate the simple interest.

### What is the Input:
- Principal → Original Amount (money invested or borrowed)
- Rate → Interest rate in percentage (e.g., 5 for 5%)
- Time → Duration in years (e.g., 2 for 2 years)

### What is the Output:
- Simple Interest amount (calculated value)
- Format: "Simple Interest is: X.XX" (float)

### Example Output:
Enter Principal: 10000
Enter Rate: 5
Enter Time: 2
Simple Interest is: 1000.0

### What are the conditions:

### What are the Conditions:
1. All three inputs must be valid numbers (no text allowed)
2. Inputs should be converted to float (decimal numbers allowed)
3. All input values must be positive (greater than 0)
4. Output should be displayed in float format

## Tools We Need:
1. Python Language (3.x or abover version)
2. Jupitor NoteBook (easy to use and debug) / VSCode IDE etc

## Data We Need:
1. float

## Skills We need
1. Create 3 variables
2. Convert Input in float
3. Use of Arithemetic Operations
   (Only * and / only)
   -Because : SI = (P x R x T) / 100 
4. Use of Input function
5. try / block to handel errors gracefully
6. print() function with f string to display out put in proper format

### Step-by-step Logical Flow:

START

1. Print Header
   - "Simple Interest Calculator"

2. Get Input from user (with try/except for error handling)
   - try:
     a. Get Principal amount
        - principal = float(input("Enter Principal: "))
     b. Get Rate of interest
        - rate = float(input("Enter Rate: "))
     c. Get Time period
        - time = float(input("Enter Time (in years): "))
   - except:
     - Print error message if input is invalid

3. Compute Simple Interest using formula
   - SI = (principal × rate × time) / 100

4. Display the final result
   - Print "Simple Interest is: {SI}"

5. Print Footer
   - "Program Completed Successfully!"

END

BEGIN SimpleInterestCalculator

    DISPLAY "Simple Interest Calculator"

    TRY
        DISPLAY "Enter Principal: "
        READ principal_input
        CONVERT principal_input TO FLOAT principal

        DISPLAY "Enter Rate: "
        READ rate_input
        CONVERT rate_input TO FLOAT rate

        DISPLAY "Enter Time (in years): "
        READ time_input
        CONVERT time_input TO FLOAT time

        COMPUTE simple_interest = (principal * rate * time) / 100

        DISPLAY "Simple Interest is: " + simple_interest
        DISPLAY "Program Completed Successfully!"

    CATCH Exception
        DISPLAY "Error: Invalid input! Please enter numeric values only."
    ENDTRY

END

```python
# ============================================
# PROGRAM 3: SIMPLE INTEREST CALCULATOR
# AUTHOR: Sarosh Khan
# DATE: September 2026
# CONCEPTS: Type Conversion, User Input,
#           Mathematical Operations, Error Handling
# ============================================

# Step 1: Print the header
print("\n" + "=" * 50)
print("\n            SIMPLE INTEREST CALCULATOR")
print("\n" + "=" * 50)

# Step 2: Get input from user via try/except
try:
    principal_amount = float(input("\nEnter principal amount: "))
    rate = float(input("\nEnter rate of interest: "))
    time = float(input("\nEnter time (in years): "))

    # Step 3: Calculate Simple Interest
    simple_interest = (principal_amount * rate * time) / 100

    # Step 4: Display final result
    print(f"\n Simple Interest is: {simple_interest:.2f}")
    print(f"   Total Amount: {principal_amount + simple_interest:.2f}")

except ValueError:
    print("\n   Invalid Input! Please enter a valid number.")

print("\n" + "=" * 50)
print("\nThank you for using Simple Interest Calculator")
print("\n" + "=" * 50)
```
