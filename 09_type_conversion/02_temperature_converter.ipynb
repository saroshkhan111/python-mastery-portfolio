Program 2: Temperature Converter (Basic)

Problem Statement: Write a program that converts temperature from Celsius to Fahrenheit.
Formula: Fahrenheit = (Celsius × 9/5) + 32
Input:
• Temperature in Celsius (float)
Output:
• Temperature in Fahrenheit

## Problem Solving Steps:


## Understand the Problem:

  - Data_Type (float)
  - Example: 0, 25, 36.5, -10, 100
  - Use Input() function to get the input from the user
  - Convert string in to the float to calculate farenheit.

## What is the Output?
   - Temperature in Fahrenheit : (Converted)
   - Formula: Fahrenheit = (Celsius × 9/5) + 32
   - Example 0°C → 32°F
   - Format "Temperature in Farenheiht: X F"

## What are the Conditions:

### Step 3: Algorithm Design:

## Step by Step Logical Flow:

START

1. Display message to user
   - "Enter temperature in Celsius: "

2. Get input from user
   - celsius_str = user input (string format)

3. Convert string to float
   - celsius = float(celsius_str)
   - (Type conversion: string → decimal number)

4. Calculate Fahrenheit using formula
   - fahrenheit = (celsius * 9/5) + 32
   - Step 4a: Multiply celsius by 9
   - Step 4b: Divide result by 5
   - Step 4c: Add 32 to the result

5. Display the calculated result with f-string
   - "Temperature in Fahrenheit: X°F"

END

### Pseudocode:

BEGIN CelsiusToFahrenheit

    DISPLAY "Enter temperature in Celsius: "
    READ celsius_str

    CONVERT celsius_str TO FLOAT celsius

    COMPUTE fahrenheit = (celsius * 9 / 5) + 32

    DISPLAY "Temperature in Fahrenheit: " + fahrenheit + "°F"

END

```python
# ============================================
# PROGRAM 2: TEMPERATURE CONVERTER
# AUTHOR: Sarosh Khan
# DATE: September 2026
# CONCEPTS: Type Conversion, User Input,
#           Mathematical Operations
# ============================================

print("\n" + "=" * 40)
print("       TEMPERATURE CONVERTER")
print("       Celsius → Fahrenheit")
print("=" * 40)

# Step 1: Get user input
try:
    celsius = float(input("\nEnter temperature in Celsius: "))
    
    # Step 2: Convert to Fahrenheit
    fahrenheit = (celsius * 9 / 5) + 32
    
    # Step 3: Display result
    print(f"\n {celsius}°C is equal to {fahrenheit:.2f}°F")

except ValueError:
    print("\n Error: Please enter a valid number!")

print("\n" + "=" * 40)
print("       PROGRAM COMPLETED SUCCESSFULLY!")
print("=" * 40)
```
