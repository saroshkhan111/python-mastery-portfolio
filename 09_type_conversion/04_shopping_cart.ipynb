### Step by Step:

1. **Subtotal = Price × Quantity**

## Pseducode
BEGIN InvoiceGenerator

    DISPLAY "========== INVOICE GENERATOR =========="

    DISPLAY "Enter Product Name: "
    READ product_name_input

    DISPLAY "Enter Price: "
    READ price_input
    CONVERT price_input TO FLOAT price

    DISPLAY "Enter Quantity: "
    READ quantity_input
    CONVERT quantity_input TO INTEGER quantity

    COMPUTE subtotal = price * quantity
    COMPUTE discount = (subtotal * 10) / 100
    COMPUTE total = subtotal - discount

    DISPLAY "---------------------------------------"
    DISPLAY "Product Name : " + product_name_input
    DISPLAY "Price        : $" + price
    DISPLAY "Quantity     : " + quantity

````python
# ============================================
# PROGRAM 4: INVOICE GENERATOR
# AUTHOR: Sarosh Khan
# DATE: September 2026
# CONCEPTS: Type Conversion, User Input,
#           Calculations, Formatting
# ============================================

# Step 1: Display header
print("\n" + "=" * 50)
print("\n                  INVOICE GENERATOR")
print("\n" + "=" * 50)

# Step 2: Get user input with error handling
try:
    product_name = input("\nEnter Product Name: ")
    price = float(input("Enter Product Price: "))
    quantity = int(input("Enter Product Quantity: "))

    # Step 3: Perform calculations
    subtotal = price * quantity
    discount = (subtotal * 10) / 100   # 10% discount
    total = subtotal - discount

    # Step 4: Display invoice
    print("\n" + "=" * 50)
    print("              INVOICE")
    print("=" * 50)
    print(f"Product Name  : {product_name}")
    print(f"Price per unit: {price:.2f}")
    print(f"Quantity      : {quantity}")
    print("-" * 50)
    print(f"Subtotal      : {subtotal:.2f}")
    print(f"Discount (10%) : {discount:.2f}")
    print("-" * 50)
    print(f"TOTAL         : {total:.2f}")
    print("=" * 50)

except ValueError:
    print("\n  Invalid Input! Please enter valid numbers.")

# Step 5: Footer
print("\n" + "=" * 50)
print("\n              INVOICE GENERATED SUCCESSFULLY!")
print("\n" + "=" * 50)
````
