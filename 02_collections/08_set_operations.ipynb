```python
#step 1: Create number set:

number_set = {101, 102, 103, 114, 225, 101, 102, 225, 225, 226,  227}


#step 2: Print the Original Set:

print(f"\nOriginal Set Values :   -> {number_set}") # duplicate values will be removed automatically in the set


#step 3: Add single element to the set using add() function
number_set.add(228)

print(f"\nAfter Adding single element to the set using add() function :\n")
print(f"Updated Set Values :  -> {number_set}\n")

#step 4: Create List and add them in the set using update() function
number_list = [229, 230, 231, 232, 233]

number_set.update(number_list)

print(f"\nAfter Updating multiple items to the set using update() function :\n")
print(f"Updated Set Values :  -> {number_set}\n")

#step 5: Remove an element from the set using remove () function
number_set.remove(102)

print(f"\nAfter Removing an element from the set using remove() function :\n")
print(f"Updated Set Values :  -> {number_set}\n")

#step 6: Discard an element from the set using discard () function
number_set.discard(103)

print(f"\nAfter discarding an element from the set using discard() function :\n")
print(f"Updated Set Values :  -> {number_set}\n")

#step 7: Perofrom the pop() operation to remove a random element from the set
popped_element = number_set.pop()
print(f"\nAfter popping a random element from the set :\n")
print(f"Updated Set Values :  -> {number_set}\n")
print(f"Popped Element :  -> {popped_element}\n")

#step 8: Create two sets to perform union , intersection , difference and symmetric operations
set_a = {1, 2, 3, 4, 5}
set_b = {1, 4, 3, 6, 9}

#step 9: Perform Union operation:

print(f"\nUnion of two sets is :  -> {set_a.union(set_b)}\n")

#step 10: Perform Intersetion operation:

print(f"\nIntersection of two sets is :  -> {set_a.intersection(set_b)}\n")

#step 11: Perform Difference operation:

print(f"\nDifference of two sets is :  -> {set_a.difference(set_b)}\n")

#step 12: Perform Symmetric difference operation:

print(f"\nSysmmetric Difference of two sets is : -> {set_a.symmetric_difference(set_b)}\n")

#step 13: check if 233 is present in the set or not

print(f"233 in number_set? {233 in number_set}\n")

#step 14: create set c

set_c = {1, 2, 3}

print(f"\n IS C subset of A? {set_c.issubset(set_a)}\n")

print(f"\nIS A Superset of C? {set_a.issuperset(set_c)}\n")

#step 15: create another set named more and add items in mumberset using updated() with for loop
more_set = {661, 662, 663, 664}

for item in more_set:
    number_set.add(item)

print(f"\nAfter adding more_set items using for loop: {number_set}\n")    


```
