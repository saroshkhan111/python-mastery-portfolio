Definition:
-----------
A dictionary in Python is a collection of key-value pairs where each unique key is connected to a value.
It is ordered, changeable, and does not allow duplicate keys.

Key Points:
-----------
* Key-Value Pairs: Each item has a key and a corresponding value, separated by a colon (:).
* Ordered: Items maintain the order in which they were added (Python 3.7+).
* Mutable: You can add, remove, or change items after creation.
* Unique Keys: Duplicate keys are not allowed; if you assign a new value to an existing key, it updates the old value.
* Dynamic: Can store any data type (strings, numbers, lists, even other dictionaries).
* Fast Lookup: Accessing a value by its key is very fast, making it ideal for large datasets.
* Syntax: Defined using curly braces {} with key-value pairs separated by commas: {"key1": value1, "key2": value2}

```python
#step 1: Create a dictionary "student"

student={
    "name":   "John Doe",
    "age":       25,
    "grade":    "B",
    "subject":  "Data-Science"
}

#step 2: Display all key of student dict
print(f"\nStudent Information in dictionary : {student} ")

#step 3: Access values using key print student name
print(f"\nStudent name using key : student[name]  ->  {student['name']}")


#step 4: Access the value of subject using get method in the student dictionary
print(f"\nAccessing Student's Subject using student.get() method : -> {student.get('subject')}")


#step 5: Add new key_value in the student dictionary -> is_student = True
student['is_student'] = True

print(f"\nAfter adding New key-value pair is_student : {student}")

#step 6: Update existing value subject data science to computer science

student['subject'] = "Computer-Science"

print(f"\nAfter Updating Subject the current dictionary is: {student}")

student ['name'] = "Ali"

print(f"\nAfter Updating Student Name the current dictionary is: {student}")

#step 7: Removing using pop(age)
student.pop("age")

print(f"\nAfter Popping age  the current dictionary is: {student}")

#step 8: Removing Using del method
del student['grade']

print(f"\nAfter using del 'grade'  the current dictionary is: {student}")

#step 9: Remove last item using popitem()
student.popitem()

print(f"\nAfter using popitem 'remove last item'  the current dictionary is: {student}")

#step 10: get all keys
print(f"\n KEYS : {student.keys()}")

#step 11: get all values
print(f"\n All Values : {student.values()}")

#step 12: get all items
print(f"\n All Items : {student.items()}")

for key, value in student.items():
    print(f" {key} : {value}")

#st 13: copy dictionary

new_dict = student.copy()

print(f"\n new dict = {new_dict} ")

#st 14: clear new dictionary

new_dict.clear()

print(f"\n new dict = {new_dict} ")    
```
