Concept:

A frozen set is an immutable version of a Python set. Once created, you cannot add, remove, or modify its elements.

```python
#step 1  Create list:

my_list = [1, 2,  3, 4, 5]

print("\n" + "="  * 50)
print("\n...............FROZENSET OPERATIONS DEMONSTRATION\n")
print("\n" + "="  * 50)

print(f"\nOriginal List : {my_list}\n")
print(f"\nFrozen_list : {frozen_list} ->  {type(frozen_list)}\n")
print(f"\nFrozen_Set : {frozen_set}-> {type(frozen_set)}\n")
#step 2 Convert list to frozen_set

frozen_list = frozenset(my_list)



#step 3 create set:

my_set = {6, 7, 8, 9, 10}


#convert my_set to frozen_set

frozen_set = frozenset(my_set)

#step 4 try to add 6 in the forzen list and demonstrate error

print("\n" + "="  * 50)
print("\n...............FROZENSET ERROR DEMONSTRATION\n")
print("\n" + "="  * 50)

try:
    frozen_list.add(6)
except AttributeError as e:
    print(f"Error :  {e}")    


#step 5 try to remove and element from frozen list 5

try:
    frozen_list.remove(5)
except AttributeError as e:
    print(f"\nError : {e}")


print("\n" + "="  * 50)
print("\n...............UNION INTERSECTION  AND  SYMEETRIC OPERATIONS\n")
print("\n" + "="  * 50)

fs_1 = frozenset([1, 2, 3, 4, 5])
fs_2 = frozenset([4, 5, 6, 7, 8, 9, 10])

print(f"\nUUNION OPERATION : {fs_1 | fs_2}")

fs_3 = frozenset({11, 12, 13, 14, 15})
fs_4 = frozenset({19, 13, 14, 15, 20})

print(f"\nINTERSTION OPERATION : {fs_3 & fs_4}")

fs_5 = frozenset((21, 22, 23, 24, 25))
fs_6 = frozenset((26, 27, 28, 29, 30))

print(f"\nDIFFERENCE OPERATION :  {fs_5 - fs_6}")

print(f"\nSYMMETRIC OPERATION :  {fs_5 ^ fs_6}")


print("\n" + "="  * 50)
print("\n...............CHECKING 229 IN FROZENLIST\n")
print("\n" + "="  * 50)

#step 6: check 229 PRESENT in the frozen list?

print (f"\nIS 229 IN frozen_list? : {229 in frozen_list}")

#step 7: Check the Subset of fs_ 1 of fs_2

print("\n" + "="  * 50)
print("\n...............CHECKING SUBSET OF FS_5  AND  FS_6 \n")
print("\n" + "="  * 50)

print(f"\n IS fs_1 is a SUBSET of fs_2? {fs_5.issubset(fs_6)}\n")


print("\n" + "="  * 50)
print("\n...............CHECKING SUPERSET OF FS_5  AND  FS_4 \n")
print("\n" + "="  * 50)


print(f"\nIS fs_5 is  a SUPERSET of fs_4? {fs_5.issuperset(fs_4)}\n")

#step 8: Print the frozen_set values throught loop
print("\n" + "="  * 50)
print("\n...............PRINTING VALUES OF MY-SET VIA FOR LOOP: \n")
print("\n" + "="  * 50)

for item in my_set:
    print(item)

#step 9: Convert frozenset to list and set

print("\n" + "="  * 50)
print("\n...............PRINTING VALUES OF MY-SET VIA FOR LOOP: \n")
print("\n" + "="  * 50)

convert_list = list(frozen_list)

print(f"AFTER CONVERSION FROZEN LIST  IN LIST {frozen_list} : {type(convert_list)}\n")

convert_set = set(my_set)

print(f"AFTER CONVERSION FROZEN SET  IN SET  {frozen_set} : {type(convert_set)}\n")


print("\n" + "="  * 50)
print("\n...............PROGRAM END: \n")
print("\n" + "="  * 50)
```
