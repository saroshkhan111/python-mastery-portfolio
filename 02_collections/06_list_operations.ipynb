```python
# step 1: Create fruit list:

fruits = ['Apple', 'Banana', 'Mango',  'Pine_Apple']


# step 2: Display the fruit list:
print("\n"  + "="  * 70)
print("\n        Original Fruit List:")
print("\n"  + "="  * 70)

#display the fruit list
print("Fruit List:")

for index, fruit in enumerate(fruits, 1):
   
    print(f"{index}. {fruit}")

# Step 3: Append 'Dates' to the fruit list
fruits.append('Dates')

#print the list after append
print(f"\nAfter Append Updated fruit list:\n")
for index, fruit in enumerate(fruits, 1):     
    
    print(f"{index}. {fruit}")

# step 3: Insert 'Grapes' in the fruit list at index 3:
fruits.insert(3, 'Grapes')

print(f"\nAfter Insert 'Grapes' Updated fruit list:\n")
for index, fruit in enumerate(fruits, 1):
    print(f"{index}. {fruit}")

# step 4: Remove 'Pine_Apple' from the fruit list:
fruits.remove('Pine_Apple')

print(f"\nAfter Remove 'Pine_Apple' Updated fruit list:\n")

for index, fruit in enumerate(fruits, 1):
    print(f"{index}. {fruit}")


# step 5: Accessing the elements of the fruit list using index:
print("\n"  + "="  * 70)
print("\n      Accessing the elements Using Index: ")
print("\n"  + "="  * 70)


print (f"\n Index = 0  : {fruits[0]}\n")

print(f"\n Index = 1  : {fruits[1]}\n") 

print(f"\n Index = 2  : {fruits[2]}\n")

print(f"\n Index = 3 : {fruits[3]}\n")


# step 5: Accessing the elements of the fruit list using index:
print("\n"  + "="  * 70)
print("\n      Slicing the first three elements:: ")
print("\n"  + "="  * 70)

fruits_slice = fruits[0:3]
print(f"\n Slicing of three elements in the fruits_list -> {fruits_slice}\n")

#step 6: Display the lenght of the fruits list:

print(f"\n Length of the fruits list is : {len(fruits)}\n")

# step 7: Display the sroted fruit list :
print("\n"  + "="  * 70)
print("\n      Sorting elements:: ")
print("\n"  + "="  * 70)
fruits.sort()
print(f"\n After Sortin the fruits Updated fruit list :\n")
print(fruits)

#step 8: Display the reversed fruit list :
print("\n"  + "="  * 70)
print("\n      Reversing elements:: ")
print("\n"  + "="  * 70)

fruits.reverse()
print(f"\n After Reversing the fruits Updated fruit list :\n")
print(fruits)

#step 9: Display the Mango and Watermelon in the list:
print("\n"  + "="  * 70)
print("\n      Displaying Mango and Watermelon in the list:: ")
print("\n"  + "="  * 70)

print(f"\n  IS Mango in the fruits lists:? {"Mango" in fruits}\n")
print(f"\n  IS Watermelon in the fruits lists:? {"Watermelon" in fruits}\n")


#step 10: Extend more fruits in the list:
print("\n"  + "="  * 70)
print("\n      Extending more fruits in the list:: ")
print("\n"  + "="  * 70)

more_fruits = ['Kiwi', 'Peach']
fruits.extend(more_fruits)

print("\n Extenede fruits lists are:\n")
print(f"\n {fruits}")


#step 11: Program Ends:
print("\n"  + "="  * 70)
print("\n      Program Ends: ")
print("\n"  + "="  * 70)
```
