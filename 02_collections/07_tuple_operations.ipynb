```python
# ============================================
# PROGRAM 7: TUPLE OPERATIONS
# AUTHOR: Sarosh Khan
# DATE: 28-Aug-2026
# ============================================

print("\n" + "=" * 50 + "\n")
print("\n....................Accessing Tuple Using Index values\n")

color_tuple = ('Red', 'Green', 'Orange', 'Black', 'Sky Blue')

# Step 2: Accessing elements
print(f"Index = 0 : -> {color_tuple[0]}")
print(f"\nIndex = 1 : -> {color_tuple[1]}")
print(f"\nIndex = 2 : -> {color_tuple[2]}")
print(f"\nIndex = 3 : -> {color_tuple[3]}")
print(f"\nIndex = 4 : -> {color_tuple[4]}")

# Step 3: Length
print("\n" + "=" * 50 + "\n")
print("\n....................Calculating Length of Tuple\n")
print(f"\nLength of Tuple is : -> {len(color_tuple)}")
print("\n" + "=" * 50 + "\n")

# Step 4: Count
print("\n....................Counting the Occurrence of an Element in the Tuple\n")
print("Occurrence of 'Red' element in the tuple is : -> ", color_tuple.count('Red'))
print("\n" + "=" * 50 + "\n")

# Step 5: Index of Black
print(f"The Index of Black is : {color_tuple.index('Black')}")

# Step 6: Immutability
print("\n" + "=" * 50 + "\n")
print("\n....................Trying to change the value of the tuple using index value\n")
try:
    color_tuple[0] = 'Yellow'
except TypeError as e:
    print(f"Error: {e}")

# Step 7: Loop with enumerate
print("\n" + "=" * 50 + "\n")
print("\n....................Printing the Tuple using for loop\n")
for index, color in enumerate(color_tuple, 1):
    print(f"{index} . {color}")

# Step 8: Tuple to List
tuple_list = list(color_tuple)
print(f"\nTuple to List : {tuple_list} -> Type : {type(tuple_list)}\n")

# Step 9: Tuple to Set
tuple_set = set(color_tuple)
print(f"\nTuple to Set : {tuple_set} -> Type : {type(tuple_set)}\n")

# Step 10: More colors
more_colors = ('Navy Blue', 'Green', 'Dark Yellow')

# Step 11: Concatenate
print("\n" + "=" * 50 + "\n")
print("\n....................Concatenating the two tuples\n")
new_color_tuple = color_tuple + more_colors
print(f"\nNew Color Tuple : {new_color_tuple} -> Type : {type(new_color_tuple)}\n")

# Step 12: Repeat
print("\n" + "=" * 50 + "\n")
print("\n....................Repeating the tuple using * operator\n")
repeat_color_tuple = color_tuple * 2
print(f"\nRepeat Color Tuple : {repeat_color_tuple} -> Type : {type(repeat_color_tuple)}\n")

# Step 13: Check existence
print("\n" + "=" * 50 + "\n")
print("\n....................Checking Red and Pink available in the tuple\n")
print(f"\n'Red' in color_tuple : {'Red' in color_tuple}")
print(f"\n'Pink' in color_tuple : {'Pink' in color_tuple}")

print("\n" + "=" * 50 + "\n")
print("\n....................Program End\n")
print("\n" + "=" * 50 + "\n")
```
