Problem: Check if a string is palindrome (reads same forwards and backwards)

Input: "madam"
Output: True (madam → madam)

Input: "hello"
Output: False (hello → olleh)

Step 1: Problem Statement:

PROGRAM: Palindrome Checker
AUTHOR: Sarosh Khan
DATE: 03-Sep-2026

PROBLEM:
Write a Python program that checks whether a given string is a palindrome or not.

DEFINITION:
A palindrome is a word, phrase, or sequence that reads the same forwards and backwards.

INPUT:
- String (e.g., "madam", "hello", "racecar")

OUTPUT:
- "madam" → Palindrome  (reads same forwards and backwards)
- "hello" → Not Palindrome  (reads "olleh" backwards)

EXAMPLES:
| Input | Reverse | Result |
|-------|---------|--------|
| madam | madam  | Palindrome  |
| hello | olleh  | Not Palindrome  |
| racecar | racecar | Palindrome  |
| python | nohtyp | Not Palindrome  |




Output Required:
----------------

Test Case 1:
Input: madam
Reverse: madam
Output: "madam is a PALINDROME "

Test Case 2:
Input: hello
Reverse: olleh
Output: "hello is NOT a PALINDROME "

Test Case 3:
Input: racecar
Reverse: racecar
Output: "racecar is a PALINDROME "

Test Case 4:
Input: python
Reverse: nohtyp
Output: "python is NOT a PALINDROME "

conditions Required:
--------------------

CONDITIONS:
1. Input should be a string
2. Compare string with its reverse
3. If equal → Palindrome
4. Else → Not Palindrome

Step 2: Requirements Analysis:

Tools Required:
---------------
- VS Code / Jupyter Notebook
- Python 3.x

Data Required:
--------------
- One string value (user input)
- Example: "madam", "hello", "racecar"

Skills Required:
----------------
1. Use input() function to get user input
2. Use string slicing [::-1] to reverse the string
3. Use if-else statement to compare original and reversed string
4. Use print() function to display the final result

Libraries Required:
-------------------
- None (All built-in Python functions)

Step 3: Algorithm Design:

Step-by-step Logical Flow:

Program Start:

1. PRINT Header:
   "PALINDROME CHECKER"

2. PROMPT user for input:
   user_input = input("Enter any string value to check palindrome: ")

3. REVERSE the string using slicing:
   reversed_string = user_input[::-1]

4. APPLY if/else condition:
   IF user_input == reversed_string:
       PRINT f"{user_input} is a PALINDROME "
   ELSE:
       PRINT f"{user_input} is NOT a PALINDROME"

5. PRINT Footer:
   "PROGRAM COMPLETED SUCCESSFULLY!"

Program End


```python
# ============================================
# PROGRAM 3: PALINDROME CHECKER
# AUTHOR: Sarosh Khan
# DATE: 03-Sep-2026
# ============================================

# Step 1: Print the header
print("\n" + "=" * 50)
print("\n.......................PALINDROME CHECKER")
print("\n" + "=" * 50)

# Step 2: Get user input and store it in a variable
user_input = input("\nEnter a word or phrase to check if it's a palindrome: ")

# Step 3: Reverse the input using slicing
reverse_string = user_input[::-1]

# Step 4: Check if the input is a palindrome or not
if user_input == reverse_string:
    print(f"\n {user_input} is a PALINDROME!")
else:
    print(f"\n {user_input} is NOT a palindrome.")

# Step 5: Print the footer
print("\n" + "=" * 50)
print("\n.......................PROGRAM COMPLETED SUCCESSFULLY!")
print("\n" + "=" * 50)
```
