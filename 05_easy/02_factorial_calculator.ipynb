PROBLEM:
Write a Python program that calculates the factorial of a given number.

INPUT:
- Integer number (non-negative)

OUTPUT:
- Factorial of the number

FORMULA:
- n! = n × (n-1) × (n-2) × ... × 1
- Example: 5! = 5 × 4 × 3 × 2 × 1 = 120

CONDITIONS:
1. Number should be non-negative
2. Use for loop or while loop
3. Display result with proper message

Step 2: Requirements Analysis:

Tools Required:
---------------
- VS Code / Jupyter Notebook
- Python 3.x (or latest version)

Data Required:
--------------
- Integer number (non-negative)
- Example: 5, 3, 7, 0

Skills Required:
----------------
1. Use input() function to take user input
2. Convert string input to integer using int()
3. Use for loop or while loop for iteration
4. Apply factorial formula: factorial = factorial * i
5. Use print() function to display result
6. Handle edge case: factorial of 0 = 1

Libraries Required:
-------------------
- None (All built-in Python functions)

Step 3: Algorithm Design:

Step-by-step Logical Flow:

Program Start:

1. PRINT Header:
   "PROGRAM TO CALCULATE FACTORIAL OF A NUMBER"

2. PROMPT user to enter an integer number:
   number = input("Enter a number: ")

3. CONVERT string to integer:
   number = int(number)

4. CALCULATE factorial:
   factorial = 1
   FOR i in range(1, number + 1):
       factorial = factorial * i

5. PRINT result:
   PRINT f"Factorial of {number} is {factorial}"

6. PRINT Footer:
   "PROGRAM COMPLETED SUCCESSFULLY!"

Program End

```python
# ============================================
# PROGRAM 2: FACTORIAL CALCULATOR
# AUTHOR: Sarosh Khan
# DATE: 02-Sep-2026
# ============================================

# Step 1: Print the Header
print("\n" + "=" * 60)
print("\n................PROGRAM TO CALCULATE FACTORIAL OF A NUMBER :")
print("\n" + "=" * 60)

# Step 2: Get the input from the user and convert it to an integer
number = int(input("\nEnter a number to calculate its factorial: "))

# Step 3: Initialize the factorial variable to 1
factorial = 1

# Step 4: Use a for loop to calculate the factorial
for i in range(1, number + 1):
    factorial *= i  # Multiply the current value of factorial by i

# Step 5: Print the factorial of the number
print(f"\nFactorial of {number} is: {factorial}\n")

# Step 6: Print the Footer
print("\n" + "=" * 60)
print("\n......................PROGRAM COMPLETED SUCCESSFULLY! :")
print("\n" + "=" * 60)
```
