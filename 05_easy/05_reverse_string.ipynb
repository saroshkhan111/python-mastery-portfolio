Problem:
Write a Python program that reverses a given string.


Problem Solving Steps:
----------------------
Problem Solving Approach

Step 1: Problem Statement:

I need to write a python program which will reverse the string

Input Required:
---------------
String Value from the user

Output Required:
----------------

user's enter : hello -> olleh

Conditions Required:
--------------------
1. Input should be string

2. use slicing [::-1]

3. no special libraries needed

4. string should be display in reverse


Step 2: Requirements Analysis:

Tools Required:
---------------
- VS Code / Jupyter Notebook
- Python 3.x

Data Required:
--------------
- String input from user
- Example: "hello", "python", "madam"

Skills Required:
----------------
1. input() function → Take user input
2. String slicing [::-1] → Reverse the string
3. print() function → Display the reversed string

Libraries Required:
-------------------
- None (All built-in Python functions)

Step 3: Algorithm Design:

Program Start:

1. PRINT Header:
   "STRING REVERSE PROGRAM"

2. PROMPT user to enter a string:
   user_input = input("Enter a string: ")

3. REVERSE the string using slicing:
   reversed_string = user_input[::-1]

4. PRINT the reversed string:
   print(f"Reversed string: {reversed_string}")

5. PRINT Footer:
   "PROGRAM COMPLETED SUCCESSFULLY!"

Program End

```python
# ============================================
# PROGRAM 5: REVERSE STRING
# AUTHOR: Sarosh Khan
# DATE: 03-Sep-2026
# ============================================

# Step 1: Print the header
print("\n" + "=" * 50)
print("\n...................STRING REVERSE PROGRAM")
print("\n" + "=" * 50)

# Step 2: Get the string input from the user
user_input = input("\nEnter a string to reverse: ")

# Step 3: Reverse the string using slicing
reversed_string = user_input[::-1]

# Step 4: Print the reversed string
print(f"\nOriginal String: {user_input}")
print(f"Reversed String: {reversed_string}")

# Step 5: Print the footer
print("\n" + "=" * 50)
print("\n...................PROGRAM COMPLETED SUCCESSFULLY!")
print("\n" + "=" * 50)
```
