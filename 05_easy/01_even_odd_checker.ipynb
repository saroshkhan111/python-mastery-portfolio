Write a Python program that checks if a number is even or odd.

Concept of Even and Odd Number:
-------------------------------

Even Numbers:
-------------
Even numbers are exactly divisible by 2 -> no reminder

Formula:
--------

n = 2k -> where k is any integer number (k = 0, ±1, ±2, ±3, ...)

Examples:  2, 4, 6, 8, 10, -4, 0, 100
For k = 1, 2, 3: n = 2, 4, 6

Odd numbers:
------------
Numbers that leave a remainder of 1 when divided by 2.
Formula:

n = 2k+1
Examples:

1, 3, 5, 7, 9, -3, 11
For k = 0, 1, 2: n = 1, 3, 5
