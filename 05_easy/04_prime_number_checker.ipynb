PROGRAM 04: PRIME NUMBER CHECKER

Problem:
Write a Python program that checks whether a given number is prime or not.

Concept of Prime Number:
------------------------

Prime number woh number hota hai jo sirf aur sirf 2 alag-alag numbers se poora divide (divide hone par remainder 0) ho sake:

Special Cases (1, 0, aur Negative Numbers):
------------------------------------------
1 prime nahi hai: Kyunki 1 ka sirf ek hi divisor hai (woh khud 1). Prime hone ke liye 2 alag divisors zaroori hain.

0 aur Negative numbers (-5, -2 etc.) prime nahi hote: Prime numbers hamesha 1 se bade integers (> 1) hote hain.


PROGRAM: Prime Number Checker

Problem Kya Hai?
------------------------------------------------
Write a Python program that checks whether a given number is prime or not.

Input Kya Hai?
------------------------------------------------
- Ek number (integer) jo user dega
- Example: 7, 10, 13, 20

Output Kya Chahiye?
------------------------------------------------
- Agar prime hai: "7 is a PRIME number"
- Agar prime nahi: "10 is NOT a PRIME number"

Conditions Kya Hain?
------------------------------------------------
1. 0 aur 1 prime nahi hain
2. 2 sab se chhota prime number hai
3. Number ko 2 se lekar (number-1) tak check karein
4. Agar kisi se divide ho gaya → Not Prime
5. Agar kisi se divide na ho → Prime

Step 2: Requirements Analysis:

Tools Required:
---------------
- VS Code / Jupyter Notebook
- Python 3.x

Data Required:
--------------
- Integer number (user input)
- Example: 7, 10, 13, 20

Skills Required:
----------------
1. input() function → Take user input
2. int() function → Convert string to integer
3. if-else condition → Check prime or not
4. for loop → Check divisibility from 2 to number-1
5. Modulo operator (%) → Check remainder
6. print() function → Display result

Libraries Required:
-------------------
- None (All built-in Python functions)

Step 3: Algorithm Design:

Program Start:

1. PRINT Header:
   "PRIME NUMBER CHECKER"

2. PROMPT user for input:
   number = int(input("Enter an integer number to check prime or not: "))

3. IF number <= 1:
   PRINT f"{number} is NOT a PRIME number "

4. ELSE:
   SET is_prime = True
   FOR i in range(2, number):
       IF number % i == 0:
           is_prime = False
           BREAK

5. IF is_prime == True:
   PRINT f"{number} is a PRIME number "
   ELSE:
   PRINT f"{number} is NOT a PRIME number "

6. PRINT Footer:
   "PROGRAM COMPLETED SUCCESSFULLY!"

Program End

```python
# ============================================
# PROGRAM 4: PRIME NUMBER CHECKER
# AUTHOR: Sarosh Khan
# DATE: 03-Sep-2026
# ============================================

# Step 1: Print Header
print("\n" + "=" * 50)
print("\n................PRIME NUMBER CHECKER")
print("\n" + "=" * 50)

# Step 2: Get input from user
number = int(input("\nEnter an integer number to check if it is prime or not: "))

# Step 3: Check if number is less than 2
if number <= 1:
    print(f"\n{number} is NOT a PRIME number ")
else:
    # Step 4: Check divisibility
    is_prime = True
    for i in range(2, number):
        if number % i == 0:
            is_prime = False
            break
    
    # Step 5: Print result
    if is_prime:
        print(f"\n{number} is a PRIME number ")
    else:
        print(f"\n{number} is NOT a PRIME number ")

# Step 6: Print Footer
print("\n" + "=" * 50)
print("\n................PROGRAM COMPLETED SUCCESSFULLY!")
print("\n" + "=" * 50)
```
