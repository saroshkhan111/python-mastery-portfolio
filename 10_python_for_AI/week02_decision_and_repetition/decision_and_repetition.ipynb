# 🚀 Python Concept: Truthiness (The Hidden Superpower)

> **Definition:** Python mein har object ke paas ek hidden boolean value (`True` ya `False`) hoti hai. Jab aap kisi object ko `if` ya `while` condition mein use karte hain, toh Python automatic check karta hai ki wo object andar se **khaali** hai ya **bhari** hai.

---

## 1. Falsy Values (Python ki nazar mein `False`)
Agar koi value "khaali" (empty) ya "zero" hai, toh Python use **`False`** treat karta hai. Yeh saari Falsy values hain:

* **Numeric Zeros:** `0`, `0.0`
* **Empty Sequences/Collections:** 
  * `""` (Empty string)
  * `[]` (Empty list)
  * `{}` (Empty dictionary)
  * `()` (Empty tuple)
* **Special Object:** `None`

---

## 2. Truthy Values (Python ki nazar mein `True`)
Inke alawa duniya ki **baaki saari values** Python mein `True` maani jaati hain:

* Non-zero numbers (`5`, `-3`, `3.14`)
* Non-empty strings (`"Hello"`, `"0"`, `"False"` bhi kyunki ye non-empty string hai!)
* Non-empty lists, dictionaries ya sets (`[1, 2]`, `{"a": 1}`)

---

## 3. Code Examples & Comparison 💻

# 📌 Concept 01: Nested If (If ke andar If)

### 💡 Ek Line Mein Samajho:
> Jab ek condition ke **True** hone par hi doosri condition check karni ho, toh hum **Nested If** use karte hain.

---

### 🥄 Spoon-Feeding Key Points:
* 🚪 **Outer `if` (Bahar wala):** Pehla darwaza hai.
* 🔐 **Inner `if` (Andar wala):** Tijori ka lock hai (yeh tabhi dikhega jab darwaza khulega).
* ❌ **Outer = False:** Python andar wale code ko dekhay ga bhi nahi (Skip kar dega).
* ✅ **Outer = True:** Python tab andar ja kar agli shart check karega.
* ⚠️ **Indentation (Spaces):** Andar wale `if` ke aage 4 spaces (Tab) lazmi hain.

---
### 🧠 Flow Chart Logic:
```text
Outer Condition (True?) 
       ├── YES ──► Inner Condition (True?) ──► [Kaam Ho Gaya ✅]
       │                  └── NO  ──────────► [Inner Else ❌]
       └── NO  ─────────────────────────────► [Outer Else 🚫]


---

### 🐍 [Code Cell] — Practical Example (ATM Machine Logic)

```python
# --- STEP 1: Variables define karein ---
card_valid = True
balance = 5000
withdraw_amount = 3000

# --- STEP 2: Nested If Logic ---
if card_valid:  # 🚪 Outer Check (Darwaza)
    print("Step 1: Card accept ho gaya! ✅")

    if balance >= withdraw_amount:  # 🔐 Inner Check (Tijori)
        print("Step 2: Transaction Successful! Cash collect karein: Rs.", withdraw_amount, "💵")
    else:
        print("Step 2: Error! Balance kam hai. ❌")

else:
    print("Step 1: Card block hai! Nikal jao. 🚫")

### 80/20 Rule (Sirf 3 Cheezein Yaad Rakho):

# 🔢🏷️ Concept 3: enumerate() — Number + Item Dono Ek Saath

### 💡 Ek Line Mein Samjho:

> `enumerate()` tumhe **do cheezein** ek saath deta hai:
> **Item ka number (index)** + **Item khud**

### Easy Level Example:

> Tum shopping list padh rahe ho:
> ❌ Purana tareeqa: "Apple... Mango... Banana..." (sirf naam)
> ✅ Naya tareeqa: "**1.** Apple, **2.** Mango, **3.** Banana" (number + naam)
> Yehi `enumerate()` karta hai!

### 80/20 Rule:
- Bina enumerate: Sirf **item** milta hai
- Enumerate ke saath: **Index (ginti)** + **Item** dono milte hain
- Default ginti **0** se shuru hoti hai
- `start=1` likho toh **1** se shuru hogi

````python
#Without Enumerate()

items = ['apple', 'banana', 'cherry', 'date']
for item in items:
    print(item, end= ", ")
````

````python
#With Enumerate()

for item_count, item in enumerate(items, start=1):
    print(f"{item_count}: {item}", end=" ")
````

# 🤝 Concept 4: zip() — Do Lists Ko Jodo

### 💡 Ek Line Mein Samjho:
> `zip()` do (ya zyada) lists ke items ko **pair** bana deta hai.
> Jaise **zip jacket** — dono sides ko jodti hai!

### 🍼 Baby Level Example:
> Tumhare paas 2 lists hain:
> - Naam: `["Ali", "Sara", "Ahmed"]`
> - Age: `[20, 22, 19]`
>

> - `("Ali", 20)`, `("Sara", 22)`, `("Ahmed", 19)`
>
> Ab har naam ke saath uski age chipak gayi! 🤝

### 80/20 Rule:
- `enumerate()` = **Ek list** → index + item
- `zip()` = **Do lists** → item + item
- Dono lists ki **length barabar** honi chahiye (warna chhoti wali pe ruk jayega)
- Result ek **tuple** milta hai har round mein

````python
items = ['laptop', 'tablet', 'smartphone', 'keyboard', 'printer']

item_count = [1, 2, 3, 4, 5]

for index, item in zip(item_count, items):
    print(f"{index}.{item}")
````

# ⚙️ `enumerate()` Ka Asal Maqsad (Under the Hood)

### 📖 Technical Definition:
> *"enumerate wraps an iterable and yields pairs of (index, item), providing an automatic counter alongside each element so you do not have to maintain one manually."*

---

### 🔍 4 Tukdon Mein Samjho:

1. **"Wraps an iterable"** 📦  
   Yeh kisi bhi List, String ya Tuple ke upar ek cover ki tarah chadh jata hai.

2. **"Yields pairs of (index, item)"** 👥  
   Har chakkar (iteration) mein 2 cheezein jodi bana kar deta hai: `(Ginti ka number, Asal cheez)`.

3. **"Automatic counter"** ⏱️  
   Ginti khud-ba-khud 0, 1, 2, 3... badhti rehti hai.

4. **"Do not have to maintain one manually"** 🛑  
   Aapko alag se `i = 0` aur `i = i + 1` likhne ka jhanjhat nahi karna padta!

---

### ⚖️ Problem vs Solution (Ise Kyun Banaya Gaya?):

| ❌ Purana / Manual Tareeqa (Bina Enumerate) | ✅ Smart Tareeqa (`enumerate` Ke Saath) |
|---|---|
| `count = 0` alag se banao | Kuch banane ki zaroorat nahi |
| Loop ke aakhir mein `count += 1` bhool jao toh bug! 🐛 | Python khud count badhata hai |
| 4 lines ka bekar code | Sirf 1 clean line! |

````python
# ============================================================
#  MANUAL COUNTER VS ENUMERATE (Comparison)
# ============================================================


print("-" * 50)
print("Manual Way")
print("-" * 50)

fruits = ['Apple', 'Banana', 'Cherry', 'Date', 'Elderberry']
count = 0 # We have to declare counter manually

for fruit in fruits:
    count += 1  # noqa: SIM113
    print(f"{count}.{fruit}")



````

````python
print("\n" + "*" * 50)
print("Enumerate Way")
print("\n" + "*" * 50)

fruits = ['Apple', 'Banana', 'Cherry', 'Date', 'Elderberry']

for index, fruit in enumerate(fruits, start=1):
    print(f"{index}.{fruit}")
````

````python
print("\n" + "*" * 50)
print("Print the items in reverse Order: ")
print("\n" + "*" * 50)

fruits = ['Apple', 'Banana', 'Cherry', 'Date', 'Elderberry']

for index, item in enumerate(reversed(fruits), start=5):
    print(f"{index}.{item}")
````

 🔄 Concept 5: While Loop — "Jab Tak Sach Hai, Chalte Raho"
 

### 💡 Ek Line Mein Samjho:

> `while` loop tab tak chalta hai jab tak **condition True** hai.
> Jaise hi condition **False** hui → loop **ruk gaya**.

> - "Kya pet bhara?" → **Nahi** → Aur khao 🍚
> - "Kya pet bhara?" → **Nahi** → Aur khao 🍚
> - "Kya pet bhara?" → **Haan** → Ruk jao! 🛑
>
> Yehi `while` loop hai!

### 80/20 Rule:
- `for` loop = **Pata hai** kitne rounds chalenge (list ke items)
- `while` loop = **Nahi pata** kitne rounds chalenge (condition pe depend)
- Condition har round se **pehle** check hoti hai
- Condition False = **0 rounds** bhi ho sakte hain

````python
# ============================================================
# 🔄 WHILE LOOP — Basic Example
# ============================================================

# Example: print the number from 1 to 5

count = 1
while count <= 5:
    print(count)
    count = count + 1


# Output:
# Count: 1
# Count: 2
# Count: 3
# Count: 4
# Count: 5

# ============================================================
#  Dry Run (Kaise Kaam Karta Hai)
# ============================================================

# Round 1: count=1 -> 1 <= 5 ? True -> print -> 1 -> count = 2 
# Round 2: count=2 -> 2 <= 5 ? True -> print -> 2 -> count = 3 
# Round 3: count=3 -> 3 <= 5 ? True -> print -> 3 -> count = 4 
# Round 4: count=4 -> 4 <= 5 ? True -> print -> 4 -> count = 5 
# Round 5: count=5 -> 5 <= 5 ? True -> print -> 5 -> count = 6 
# Round 6: count=6 -> 6 <= 5 ? False -> exit loop
````

````python
# ============================================================
# ⚖️ FOR vs WHILE — Kab Kya Use Karo?
# ============================================================

# For loop pata hai kitni bar loop chalega
print("===FOR LOOP FIXED ROUNDS)===")
fruits = ["apple", "banana", "cherry"]
for num, fruit in enumerate(fruits, start = 1):
    print(f"{num}. {fruit}")
````

````python
#=============================================================
 # WHILE NHAI PATA KAB TAK CHALE GA 
# ============================================================

print("\n===WHILE LOOP UNKNOWN ROUNDS===")

password = " "
while password != "1234":
    password = input("Enter Correct Password: ")
    #Kitni bar poche ga pata nahi
    #Jab tak sahi passward nahi mile ga chalta rahe ga
print("Correct Password Entered!")    
````

# 🔄 While Loop + Break (Sirf 3 Parts)

### Yeh code 3 cheezein karta hai:

| Part | Kya Hai? | Real Life Example |
|------|----------|-------------------|
| **Part 1** | Variables set karo | Race shuru karne se pehle position set karo |
| **Part 2** | While condition lagao | "Jab tak 100 nahi pohanche, bhaagte raho" |
| **Part 3** | Break (Emergency Brake) | "20 rounds ho gaye? Ruk jao warna thak jaoge!" |

````python
# PART 1 : VARIABLES

# Before starting the loop set the initial condition for the variable

user_marks = 60  # initial point of marks
counter = 0      # initial count of iterations

print(f"Starting Marks: {user_marks}, Counter = {counter}")

````

````python
#Part 2 : LOOP AND REPETITION




user_marks = 60
counter = 0
print(f"Starting Marks: {user_marks}, Counter = {counter}\n")

while user_marks < 100:
    user_marks += 5  # increment marks by 5 in each iteration
    counter += 1     # increment counter by 1 in each iteration
    #Round 1: 65
    #Round 2: 70
    #.........
    #When marks reach to 100 < 100 loop will be terminate 

    print(f"Final Marks: {user_marks}, Counter = {counter}")
````

````python
# Break Statement Example

# ============================================================
#  PART 3: BREAK (Emergency Brake — Safety Net)
# ============================================================

# Problem if marks does not reach 100, we can stuck in the infinite loop.

user_marks = 60

counter = 0

while user_marks < 100:
    print(f"Round {counter+1}: Marks = {user_marks}")

    user_marks += 5
    counter +=1


    if user_marks > 20:
        print(f"Round Complete for Round {counter}")
        break

print("Loop Ended")
````

### 🔍 Dry Run (Samjho Kaise Chala):

| Round | user_marks | counter | marks < 100? | counter > 20? | Kya Hua? |
|-------|-----------|---------|-------------|--------------|----------|
| 1 | 60 | 1 | ✅ True | ❌ No | Print 60 |
| 2 | 65 | 2 | ✅ True | ❌ No | Print 65 |
| 3 | 70 | 3 | ✅ True | ❌ No | Print 70 |
| ... | ... | ... | ... | ... | ... |
| 8 | 95 | 8 | ✅ True | ❌ No | Print 95 |
| 9 | 100 | 9 | ❌ **False** | ❌ No | **LOOP RUK GAYA!** 🛑 |

> ⚠️ Is case mein break ki zaroorat nahi padi kyunki marks 100 pohanch gaye.
> Break tab kaam aata hai jab marks kabhi 100 na pohanchein (safety net!).

Lekin tumhari ek shart hai: Jaise hi koi item dukan par NAHI MILEGA, tum gussay mein dukaan chhor kar bahar nikal jaoge (break)!

🔄 Round 1:
- Item: 'laptop'
- D-Mart mein hai? -> HAAN ✅ ('laptop' dmart_items mein hai)
- Output: "Bought laptop" print hua!

🔄 Round 2:
- Item: 'mouse'
- D-Mart mein hai? -> NAHI ❌ ('mouse' dmart_items mein NAHI hai)
- Code gaya ELSE block mein -> wahan mil gaya BREAK! 🛑
- BREAK ne poora FOR LOOP wahi KHATAM kar diya!

🚫 Round 3 & 4:
- 'keyboard' aur 'monitor' tak Python POHANCHA HI NAHI!
- Kyunki Round 2 mein hi BREAK ne loop ko maar diya tha.

````python
# ============================================================
# 🛒 D-MART SHOPPING LOGIC 
# ============================================================

my_items = ['laptop','mouse',  'keyboard', 'monitor']

dmart_items = ['laptop' ,'keyboard', 'monitor']

for item in my_items:   # collect list of ll my items
    if item in dmart_items:  # check if my_items is available in demart_items
        print(f"Bought: {item}") # if Available bought 'Laptop'
    else:
        break   # break 
````

💡 Is Code Ka Asal Maqsad:

Is code mein break ki jagah continue use hua hai.

* 🛑 break = Jaise hi item nahi mila → Dukan chhor kar baahar nikal jao (loop khatam).
*  ⏭️ continue = Jaise hi item nahi mila → Is item ko chhoro aur aglay item par jao (loop chalti rahegi). 

````python
# SHOPPING WITH CONTINUE (Agla Item Skip Karo)

items = ['laptop', 'mouse', 'keyboard', 'monitor']

dmart_items = ['mouse', 'monitor', 'laptop']

for item in items:
    if item in dmart_items:
        print(f"Bought: {item}")
        continue  # skip to the next item if available in dmart_items
    else:
        print(f"{item} is not available in D-Mart")
````

# 🔢 Odd Numbers Print Karo (`continue` Ka Magic)

### 💡 Ek Line Mein Samjho:
> `range(1, 7)` se 1 se 6 tak ginti milegi.
> `i % 2 == 0` se Even (Juft) numbers check honge.
> Jab bhi Even number aayega, `continue` usse **skip (chhor)** dega.
> Result: Sirf **Odd (Fard) numbers** `1, 3, 5` print honge!

---

### 🧩 2 Important Terms:
1. **`i % 2 == 0` (Modulo Operator):**
   * Agar number 2 se poora divide ho jaye (remainder 0 bache) → Even Number (2, 4, 6)
2. **`continue`:**
   * Agla code mat chalao, seedha **aglay number par jao**!

````python
for num in range(1, 7):  # >>> print 1 -> 6
    if num % 2 == 0:     #   1 % 2 = 1,     
        continue         #   2, 4, 6    
    else:
        print(num)  #1, 3, 5
````

💡 Summary (Khulasa)

any() aur all() Python ke built-in functions hain jo kisi list ya collection ke saare elements ko ek saath test karte hain:

1. any(): Jab aapko dekhna ho ke list mein kam se kam ek condition/item True hai ya nahi.
2. all(): Jab aapko dekhna ho ke list ke saare ke saare conditions/items True hain ya nahi.

📌 Key Bullet Points:

1. any() = Koi Ek Bhi True Ho: Agar collection mein 100 mein se sirf 1 item bhi True hua, toh any() aapko True dega. Sab False honge tabhi False dega.

2. all() = Saare Ke Saare True Hon: Agar list ka har ek item True hoga, tabhi all() aapko True dega. Agar ek bhi False mil gaya, toh yeh False de dega.

3. Truthy aur Falsy Values:
   - Truthy (Sahi): True, Non-zero numbers (1, 2), Non-empty strings ("Ali").
   - Falsy (Galat): False, Zero (0), Empty string (""), Empty list ([]).

 80/20 Rule: Industry mein all() ka use form validation (dekhna ke saare fields fill hue ya nahi) aur any() ka use system checks (kya koi ek system fail hua?) ke liye bohot hota hai.  

````python
#Example 1: Basic Comparison between any() and all()

conditions = [True, False, True]

print(any(conditions))
print(all(conditions))
````

````python
# Example 2: Real-World Form Validation with all()

name_field = True
email_field = True
age_field = False

form_ready = all([name_field, email_field, age_field])

print(f"form_ready {form_ready}")
````

````python
# Example 3: Checking Offers/Discounts with any()

has_student_card = True
has_promo_code = False

get_discount = any([has_student_card, has_promo_code])

print(f"get_discount {get_discount}")
````
