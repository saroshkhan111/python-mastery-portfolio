Class __16 if/elif/else & Truthiness

A conditional statement is just like making a choice in real life.
Think of it as a rule for your computer: "If this happens, do A. Otherwise, if that happens, do B. If nothing happens, do C."

- if: Tries the first option.
- elif: Tries the next option if the first one didn't work
- else: The final backup plan if nothing else worked.

Key takeaway: No matter how many options you give, the computer will only pick one of them to run.

````python
#Example
#if basic shape
is_raining = True

if is_raining:
    print("Take an Umberella")
````

````python
#Example 2

is_not_raining = True

if is_not_raining == False:
    print("Go Outside")
else:
    print("Don't go outside")   
````

````python
if_false = False # this block won't execute becuase it only run true value
if if_false:
    print("if block is executing")

print("block is executing...")

````

````python
if_true = True # Now we have true v
if if_true:
    print("Now if block is executing....")
````

# 🚀 Python Concept: Truthiness (The Hidden Superpower)

> **Definition:** Python mein har object ke paas ek hidden boolean value (`True` ya `False`) hoti hai. Jab aap kisi object ko `if` ya `while` condition mein use karte hain, toh Python automatic check karta hai ki wo object andar se **khaali** hai ya **bhari** hai.

---

## 1. Falsy Values (Python ki nazar mein `False`)
Agar koi value "khaali" (empty) ya "zero" hai, toh Python use **`False`** treat karta hai. Yeh saari Falsy values hain:

* **Numeric Zeros:** `0`, `0.0`
* **Empty Sequences/Collections:** 
  * `""` (Empty string)
  * `[]` (Empty list)
  * `{}` (Empty dictionary)
  * `()` (Empty tuple)
* **Special Object:** `None`

---

## 2. Truthy Values (Python ki nazar mein `True`)
Inke alawa duniya ki **baaki saari values** Python mein `True` maani jaati hain:

* Non-zero numbers (`5`, `-3`, `3.14`)
* Non-empty strings (`"Hello"`, `"0"`, `"False"` bhi kyunki ye non-empty string hai!)
* Non-empty lists, dictionaries ya sets (`[1, 2]`, `{"a": 1}`)

---
