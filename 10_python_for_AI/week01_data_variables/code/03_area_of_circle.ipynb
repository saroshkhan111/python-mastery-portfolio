## Problem Statement:

Write a Python program to calculate and display the area of a circle with a radius of 7

units. Use the formula Area = pi times r^2, where pi = 3.14159.

1. Input: Radius (r = 7)
2. Output: Area of the circle
3. Formula: area = 3.14159 * radius ** 2

```python
# step 1: create a variable and store the value of radius

radius = 7

# step 2: Process the Calculation
PI = 3.14159


area = PI * (radius ** 2) 

# step 3: Display the area of a circle

print(f"Area of the the Circle is: {area:.2f}")
```

## Problem 2: Student Percentage Calculation

Write a Python program to calculate and display the percentage score of a student.

**Given Data:**
* **Total Marks:** 500
* **Obtained Marks:** 425

**Formula:**
$$
\text{Percentage} = \left( \frac{\text{Obtained Marks}}{\text{Total Marks}} \right) \times 100
$$

Compute the percentage and display the result formatted nicely on the screen.

```python
# step 1: create two variables:

total_marks = 500
obtained_marks = 425

# step 2: compute the total_percentage

percentage_score = (obtained_marks/total_marks)*100

# step 3: print the final marks on the screen

print(f"PERCENTAGE OF STUDENT SCORE :  -> {percentage_score:.2f}%")
```

## Problem 4: Even Division / Remainder Check

Write a Python program to check whether dividing `2826` by `4` leaves a remainder of `0`.

**Given Data:**
* **Dividend (Number):** 2826
* **Divisor:** 4

**Task:**
Use the Modulus operator (`%`) to find the remainder and verify if `remainder == 0`.

```python
# step 1: create and initialize variables:

number_1 = 2826
number_2 =  4

# step 2: compute the remainder:

remainder = (number_1  %  number_2)

# step 3: display the final result with if/else statement:

if remainder == 0:
    print(f"{number_1} is completly divisble no remainder left")
else:
    print(f"{number_1} is not completly divisible {remainder} left")
```
