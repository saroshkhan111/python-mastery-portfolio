## What is a Variable?:

A variable is a named container that holds data in computer memory. In Python, variables are dynamically typed—you do not need to explicitly declare the data type.

### Creating variables
1. student_count = 45       # Integer
2. learning_rate = 0.01     # Float
3. dataset_name = "MNIST"   # String
4. is_trained = True        # Boolean

### How Variables Work in Python:

Variables act as references (pointers) to objects in memory, rather than holding the actual value directly.

x = 10  # 'x' points to the integer object 10
y = x   # 'y' points to the same object 10

### Mandatory Python Syntax Rules:

1. Allowed Characters: Only letters (a-z, A-Z), numbers (0-9), and underscores (_).

2. Starting Character: Must start with a letter or an underscore. Cannot start with a digit.

3. Case Sensitivity: total_loss, Total_Loss, and TOTAL_LOSS are three separate variables.

4. Reserved Keywords: Cannot use Python's built-in keywords (e.g., for, class, import, def, return).

### Example:

# VALID
1.   user_age = 24
2.   _cache = {}
3.     score2 = 95.0

# INVALID (SyntaxError)
1. 2nd_player = "Alex"    # Starts with a digit
2. user-name = "Jordan"   # Hyphens are subtraction operators
3. class = "Math"         # 'class' is a reserved keyword

# Operator Precedence:

BODMAS is a rule used to determine the correct order of operations in mathematics

BODMAS stand for:

- B - Brackets
- O - Orders (powers, exponent, roots)
- D - Divsion
- M - Multiplication
- A - Addition
- S - Subtraction

2 + 2 - 3 = 4 - 3 = 1
2 + 2 - 3 = 2 - 1 = 1

2 + 2 * 3 = 2 + 6 = 8

2 * (2 + 3) = 2 * 5 = 10



## Order to follow
1. Solve anything   inside Brackets first.
2. Calculate Orders (e.g., squares, cubes, square roots).
3. Perform Division and Multiplication from left to right (whichever comes first).
4. Perform Addition and Subtraction from left to right    (whichever comes first).

Important: Division is not always before multiplication, and addition is not always before subtraction. They have the same precedence, so evaluate them from left to right.

````python
print(2 + 2 * 3) # 8 following BODMAS RULE

# y = mx + c --> it uses mostly in linear regression

m = 5

x = 1

c = 3

y = (m*x) + c
````

Technical definition - Keyword: A keyword (reserved word) is a word that Python reserves for its own syntax, such as if, for, or def.
Because the language already gives these words a special meaning, you cannot use them as variable, function, or class names.

````python
int = 5 ## wrong way

num5 = 5 ## right way

````

````python
# Common Mistake: Reserved keywords cannot be used as function names
# and()  # This would cause: SyntaxError: invalid syntax

# 'and' is a reserved keyword - it's a logical operator, not a function
# Example of correct usage:
result = True and False
print(result)  # False
````

````python
# Common Mistake: Reserved keywords cannot be used as variable names
# or = 7  # This would cause: SyntaxError: invalid syntax

# 'or' is a reserved keyword - it's a logical operator
# Correct way:
result_value = 7
print(result_value)
````

````python
# Common Mistake: Reserved keywords cannot be used as class names
# class and:
#     pass
# This would cause: SyntaxError: invalid syntax

# 'and' is a reserved keyword and cannot be used as a class name
# Correct way:
class MyClass:
    pass

print("Class created successfully")
````
