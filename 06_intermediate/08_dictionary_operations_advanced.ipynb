# 📚 Dictionary in Python:
--------------------------

## 🎯 Dictionary Kya Hai?:
---------------------------

Dictionary ek **data structure** hai jo data ko **key-value pairs** mein store karta hai

Soch aise:
- Real life mein **phone book** hota hai → Naam (key) + Number (value)
- Real life mein **English-Urdu dictionary** hoti hai → Word (key) + Meaning (value)

Python mein bhi **same concept** hai!
-------------------------------------
## 🔧 Syntax (Banane Ka Tareeqa)

dictioanry name = {key 1: value1,  key 2: , value2, key3:, value3}

**3 Zaroori Cheezein:**
| Part | Matlab | Example |
|------|--------|---------|
| `{}` | Curly braces (dictionary brackets) | `{ }` |
| `key` | Unique naam (jaise naam, roll no) | `"Ali"`, `1`, `"city"` |
| `value` | Data (jaise marks, age, address) | `85`, `20`, `"Lahore"` |

## 💡 Real Life Examples:
-------------------------

# Student ke grades
grades = { "Ali": 85, "Sara": 90, "Ahmed": 78 }

# Phone book
phone = { "Ali": "0300-1234567", "Sara": "0321-7654321" }

## ⚡ Important Rules:
----------------------
## ⚡ Important Rules

| Rule | Example |
|------|---------|
| Keys **unique** honi chahiyein | ✅ `{ "Ali": 85, "Ali": 90 }` → Ali: 90 (pehli value replace) |
| Keys **immutable** hon (change na ho) | ✅ String, Number, Tuple |
| Values **kuch bhi** ho sakti hain | ✅ List, Dict, Number, String |
| Comma `,` lagana zaroori hai | ✅ `{ "a": 1, "b": 2 }` |
| Colon `:` key aur value ke beech | ✅ `{ "key": value }` |


## 🔄 List vs Dictionary

| List | Dictionary |
|------|------------|
| Ordered collection | Key-value pairs |
| Index se access (0, 1, 2) | Key se access |
| `[ ]` brackets | `{ }` brackets |
| Duplicates allowed | Keys unique honi chahiyein |



## 🔍 Dictionary Mein Data Access Kaise Karein?:
------------------------------------------------

### Method 1: Square Brackets `[]

grades = { "Ali": 85, "Sara": 90 }
print(grades["Ali"])   # Output: 85

**Problem:** Agar key nahi mili toh **Error** aayega! ❌
print(grades["Ahmed"])  # KeyError! ❌

### Method 2: `.get()` Method (Safe Tareeqa ✅)
grades = { "Ali": 85, "Sara": 90 }
print(grades.get("Ali"))        # Output: 85
print(grades.get("Ahmed"))     # Output: None (Error nahi aata!)
print(grades.get("Ahmed", 0))  # Output: 0 (default value)

**Kab kaunsa use karein?**
| Situation | Best Method |
|-----------|-------------|
| Key **pakka milni** hai | `dict[key]` |
| Key **ho bhi sakti hai, nahi bhi** | `dict.get(key)` |
| Default value chahiye | `dict.get(key, default)` |

## ✏️ Dictionary Mein Changes Kaise Karein:

**`del` keyword:**
```python

**`clear()` method (sab hatana):**
```python

### 🎯 pop() vs del — Farq Kya Hai?

| Feature | `pop()` | `del` |
|---------|---------|-------|
| Removed value return karta hai? | ✅ Haan | ❌ Nahi |
| Syntax | `dict.pop(key)` | `del dict[key]` |
| Default value de sakte hain? | ✅ `dict.pop(key, default)` | ❌ Nahi |

## 🪆 Nested Dictionary (Dictionary Ke Andar Dictionary)

## Key Points to Remember

| Operation | What it does | Returns something? |
|-----------|--------------|-------------------|
| `dict[key]` | Gets the value | Yes |
| `dict.get(key)` | Gets the value safely | Yes |
| `dict[key] = value` | Adds or changes | No |
| `dict.pop(key)` | Removes the key | Yes (the value) |
| `del dict[key]` | Removes the key | No |
| `dict.keys()` | Shows all keys | Yes |
| `dict.values()` | Shows all values | Yes |
| `dict.items()` | Shows all pairs | Yes |
| `key in dict` | Checks if key exists | True or False |
| `dict.copy()` | Makes a copy | Yes (new dict) |
| `dict.clear()` | Removes everything | No |

````python
# ============================================================
# PROGRAM 8: DICTIONARY ADVANCED OPERATIONS
# Author: Sarosh Khan
# Date: Sep 2026
# ============================================================

print("=" * 60)
print("DICTIONARY ADVANCED OPERATIONS")
print("=" * 60)


# Step 1: Create a nested dictionary of students
students = {
    "STU001": {"name": "Ali",   "grade": 85, "age": 20, "city": "Lahore"},
    "STU002": {"name": "Sara",  "grade": 90, "age": 21, "city": "Karachi"},
    "STU003": {"name": "Ahmed", "grade": 78, "age": 19, "city": "Islamabad"}
}

print("\n1. Created Dictionary:")
print(students)
print("Data Type:", type(students))


# Step 2: Access values using square brackets
print("\n2. Access using [] :")
print("Ali's name  :", students["STU001"]["name"])
print("Sara's grade:", students["STU002"]["grade"])
print("Ahmed's age :", students["STU003"]["age"])


# Step 3: Access values using get() method
print("\n3. Access using .get() :")
print("Ahmed's city  :", students["STU003"].get("city"))
print("Ali's phone   :", students["STU001"].get("phone", "Not Found"))
print("Shakir's data :", students.get("STU004", "Student Not Found"))


# Step 4: Add a new student
print("\n4. Add new key-value pair:")
print("Before adding:", list(students.keys()))

students["STU004"] = {"name": "Shakir", "grade": 95, "age": 22, "city": "Peshawar"}

print("After adding :", list(students.keys()))
print("New student  :", students["STU004"])


# Step 5: Update an existing value
print("\n5. Update existing value:")
print("Ahmed's age before:", students["STU003"]["age"])

students["STU003"]["age"] = 22

print("Ahmed's age after :", students["STU003"]["age"])


# Step 6: Remove item using pop()
print("\n6. Remove using pop():")
removed_city = students["STU003"].pop("city")
print("Removed value      :", removed_city)
print("Ahmed after pop    :", students["STU003"])


# Step 7: Remove item using del
print("\n7. Remove using del:")
del students["STU002"]["grade"]
print("Sara after del     :", students["STU002"])


# Step 8: Get all keys
print("\n8. Get all keys:")
print("Main dictionary keys:", students.keys())
print("Ali's keys          :", students["STU001"].keys())


# Step 9: Get all values
print("\n9. Get all values:")
print("Ali's values:", students["STU001"].values())


# Step 10: Get all items
print("\n10. Get all items:")
print("Ali's items:", students["STU001"].items())


# Step 11: Check if a key exists
print("\n11. Check if key exists:")
print("Is STU001 present?", "STU001" in students)
print("Is STU009 present?", "STU009" in students)


# Step 12: Loop through the dictionary
print("\n12. Loop through dictionary:")
for key, value in students.items():
    print(f"\n{key} :")
    for inner_key, inner_value in value.items():
        print(f"   {inner_key} : {inner_value}")


# Step 13: Nested dictionary operations
print("\n13. Nested dictionary operations:")
print("Ali's grade before:", students["STU001"]["grade"])

students["STU001"]["grade"] = 95

print("Ali's grade after :", students["STU001"]["grade"])


# Step 14: Copy the dictionary
print("\n14. Copy dictionary:")
copy_dict = students.copy()
print("Original keys:", list(students.keys()))
print("Copy keys    :", list(copy_dict.keys()))


# Step 15: Clear the original dictionary
print("\n15. Clear dictionary:")
students.clear()
print("Original after clear:", students)
print("Copy is still safe  :", list(copy_dict.keys()))


print("\n" + "=" * 60)
print("ALL OPERATIONS COMPLETED")
print("=" * 60)
````
