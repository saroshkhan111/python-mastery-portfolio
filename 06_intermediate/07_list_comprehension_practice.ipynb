Program Start:

1. PRINT Header: "LIST COMPREHENSION PRACTICE"

2. TASK 1 – SQUARES OF 1 TO 10:
   FOR each number x FROM 1 TO 10:
       CREATE square = x * x
       ADD square TO list
   PRINT the list

3. TASK 2 – CUBES OF 1 TO 10:
   FOR each number x FROM 1 TO 10:
       CREATE cube = x * x * x
       ADD cube TO list
   PRINT the list

4. TASK 3 – EVEN NUMBERS FROM 1 TO 20:
   FOR each number x FROM 1 TO 20:
       IF x % 2 == 0:
           ADD x TO list
   PRINT the list

5. TASK 4 – ODD NUMBERS FROM 1 TO 20:
   FOR each number x FROM 1 TO 20:
       IF x % 2 != 0:
           ADD x TO list
   PRINT the list

6. TASK 5 – LENGTH OF EACH WORD:
   CREATE word_list = ["apple", "banana", "mango"]
   FOR each word IN word_list:
       CREATE length = LEN(word)
       ADD length TO list
   PRINT the list

7. TASK 6 – UPPERCASE OF EACH WORD:
   FOR each word IN word_list:
       CREATE upper_word = word.upper()
       ADD upper_word TO list
   PRINT the list

8. TASK 7 – DIVISIBLE BY 3 (1 TO 30):
   FOR each number x FROM 1 TO 30:
       IF x % 3 == 0:
           ADD x TO list
   PRINT the list

9. TASK 8 – SQUARE OF EVEN NUMBERS (1 TO 10):
   FOR each number x FROM 1 TO 10:
       IF x % 2 == 0:
           CREATE square = x * x
           ADD square TO list
   PRINT the list

```python
#step 1:  print the header
print("\n" + "=" * 50)
print("\n...............LIST COMPREHENSION PRACTICE")
print("\n" + "=" * 50)


#step 2: print the square of numbers from 1 to 10 using list comprehension
square = [x**2 for x in range(1,11)]

print(f"\nSquare Result : {square}    Data-type : ->  {type(square)}")

print(f"\nOriginal list: {list(range(1,11))}")

#step 3: print the cube list of numbers 1 to 10 using list comprehenstion

cube = [x**3 for x in range(1,11)]

print(f"\nCube Result : {cube}  Data-type : -> {type(cube)}")

print(f"\nOriginal list: {list(range(1,11))}")

#step 4: print the even numbers from 1 to 20 using list comp[rehension]

even_numbers = [even  for even  in range(1,21) if even % 2 == 0]
print(f"\nEven Number Result : {even_numbers}  Data-Type -> {type(even_numbers)}")

with_out_even_numbers = [even  for even  in range(1,21)]
print(f"\nOriginal List : {with_out_even_numbers}")

#step 5: print odd numbers using list comprehension
odd_numbers = [odd  for  odd in range(1,21) if odd % 2 == 1]
print(f"\nOdd numbers Result : {odd_numbers}  Data-Type -> {type(odd_numbers)} ")

with_out_odd = [odd for odd in range(1,21)]
print(f"\nOriginal Numbers:  {with_out_odd}")

#step 6: count the lenght of word

word_list = [len(word)  for  word in ["apple", "banana", "mango"]]

print(f"\nlenght of words : {word_list}  Data_Type -> {type(word_list)} ")

word_list = [word  for  word in ["apple", "banana", "mango"]]

print(f"\nWithout Lenght : {word_list}")

#step 7: print the uppercase() case of each word
upper_case = [word.upper()  for  word in ["apple", "banana", "mango"]]

print(f"\nUpper Case word {upper_case} Data_Type -> {type(upper_case)}")

#step 8: print the divisbility by 3 (1 to 30)
divisisble_by_3 = [x  for  x in range(1, 31) if x % 3 == 0]

print(f"\nDivisibility By 3 : {divisisble_by_3} Data_Type -> {type(divisisble_by_3)}")

#step 9: print the square of even_number (1 to 10)

even_square = [x ** 2  for  x  in range(1, 11) if x % 2 == 0]
print(f"\neven_numbers square '1 to 10 ' {even_square}  Data_Type -> {type(even_square)}")

#step 10: Print the footer
print("\n" + "=" * 50)
print("\n...............LIST COMPREHENSION PRACTICE COMPLETED!")
print("\n" + "=" * 50)
```
